<div class="eda-container">
    <div class="eda-title">🔁 تحويل التسميات إلى صيغة YOLO</div>
    <div class="eda-text-box">
        تهدف هذه الخلية إلى توضيح
        <span class="eda-highlight">آلية تحويل إحداثيات مربعات الاحتواء</span>
        من التنسيقات الشائعة (مثل CSV أو Pascal VOC)
        إلى الصيغة المطلوبة من قبل نموذج YOLO.
    </div>
    <div class="eda-text-box">
        تعتمد صيغة YOLO على تمثيل كل كائن باستخدام
        <span class="eda-highlight">إحداثيات نسبية</span>
        تشمل مركز الصندوق (x, y) والعرض والارتفاع،
        وجميعها مُطبّعة بالنسبة لأبعاد الصورة.
    </div>
    <div class="eda-text-box">
        تتضمن الخلية دالة مساعدة لتحويل الإحداثيات،
        بالإضافة إلى مثال عملي يوضح كيفية
        تحويل ملف تسميات بصيغة CSV
        إلى ملفات نصية منفصلة
        <span class="eda-highlight">(.txt)</span>
        متوافقة مع YOLO.
    </div>

</div>

<div class="eda-note">
    ℹ️ هذه الدالة توضيحية ويمكن تعديلها لتتناسب مع تنسيق التسميات المستخدم في المشروع.
</div>

</div>


```python
# خلية بايثون — مثال دالة مساعدة لتحويل تسميات من التنسيقات الشائعة إلى صيغة YOLO
# (هذه دالة توضيحية؛ عدّلها بحسب تنسيق التسميات لديك)
from pathlib import Path
from PIL import Image

def convert_bbox_to_yolo(box, img_w, img_h):
    # box = [xmin, ymin, xmax, ymax]
    xmin, ymin, xmax, ymax = box
    x_center = ((xmin + xmax) / 2) / img_w
    y_center = ((ymin + ymax) / 2) / img_h
    width = (xmax - xmin) / img_w
    height = (ymax - ymin) / img_h
    return x_center, y_center, width, height

# مثال: تحويل ملف CSV إلى ملفات txt يولو
def csv_to_yolo(csv_path, images_dir, labels_out_dir):
    import csv
    labels_out_dir = Path(labels_out_dir)
    labels_out_dir.mkdir(parents=True, exist_ok=True)
    with open(csv_path, newline='', encoding='utf-8') as f:
        reader = csv.DictReader(f)
        for row in reader:
            img_file = Path(images_dir)/row['image']
            if not img_file.exists():
                continue
            img = Image.open(img_file)
            w,h = img.size
            xmin = float(row['xmin']); ymin=float(row['ymin'])
            xmax = float(row['xmax']); ymax=float(row['ymax'])
            cls = int(row.get('class_id',0))
            x,y,wd,ht = convert_bbox_to_yolo([xmin,ymin,xmax,ymax], w,h)
            out = labels_out_dir / (img_file.stem + '.txt')
            with open(out,'a', encoding='utf-8') as fo:
                fo.write(f"{cls} {x} {y} {wd} {ht}\n")
```
