<div class="eda-container">
    <div class="eda-title">🖼️ توحيد أبعاد الصور (Image Resizing)</div>
    <div class="eda-text-box">
        تهدف هذه الخلية إلى تنفيذ خطوة <span class="eda-highlight">توحيد حجم الصور</span>
        داخل مجلد التدريب، وهي من أهم خطوات المعالجة المسبقة قبل إدخال البيانات
        إلى نموذج التعلم العميق.
    </div>
    <div class="eda-text-box">
        يتم المرور على جميع الصور داخل المجلد المحدد،
        ثم إعادة تحجيم كل صورة إلى أبعاد ثابتة
        <span class="eda-highlight">(300 × 300)</span>
        لضمان أن جميع العينات تمتلك نفس الأبعاد.
    </div>
    <div class="eda-text-box">
        توحيد الأبعاد ضروري لأن:
        <ul style="margin-right: 20px;">
            <li>نماذج الشبكات العصبية تتطلب مدخلات ذات أبعاد ثابتة</li>
            <li>اختلاف أحجام الصور قد يؤدي إلى أخطاء أثناء التدريب</li>
            <li>يساعد على استقرار الأداء وتحسين سرعة المعالجة</li>
        </ul>
    </div>
    <div class="eda-text-box">
        بعد إعادة التحجيم، يتم <b>حفظ الصور فوق النسخ الأصلية (Overwrite)</b>،
        مما يقلل من استهلاك المساحة ويجعل البيانات جاهزة مباشرة
        للاستخدام في المراحل اللاحقة.
    </div>
</div>

<div class="eda-note">
    ℹ️ اختيار حجم 300×300 يمثل توازنًا بين الحفاظ على التفاصيل وتقليل العبء الحسابي أثناء التدريب.
</div>

</div>


```python
# Resize all images in datasat orginal/Train to 300x300 (overwrite originals)

import cv2
from pathlib import Path
from tqdm import tqdm

TRAIN_PATH = Path("datasat orginal/Train")
TARGET_SIZE = (300, 300)

# التأكد أن المسار موجود
assert TRAIN_PATH.exists(), "Train path not found!"

image_extensions = {".jpg", ".jpeg", ".png"}

# جمع جميع الصور
image_paths = [
    img_path
    for class_dir in TRAIN_PATH.iterdir() if class_dir.is_dir()
    for img_path in class_dir.iterdir()
    if img_path.suffix.lower() in image_extensions
]

print(f"Total images to resize: {len(image_paths)}")

# تغيير الحجم واستبدال الصور
for img_path in tqdm(image_paths, desc="Resizing images"):
    img = cv2.imread(str(img_path))
    if img is None:
        continue

    img_resized = cv2.resize(img, TARGET_SIZE)
    cv2.imwrite(str(img_path), img_resized)

print("✅ All images resized to 300x300 and overwritten successfully.")
```

<div class="eda-intro">

<div class="eda-container">
    <div class="eda-title">🧹 Step 7: إزالة الضوضاء من الصور (Image Denoising)</div>
    <div class="eda-text-box">
        تهدف هذه الخلية إلى تنفيذ خطوة <span class="eda-highlight">إزالة الضوضاء (Denoising)</span>
        من الصور، وهي خطوة مهمة لتحسين جودة البيانات البصرية
        قبل إدخالها إلى نموذج التعلم العميق.
    </div>
    <div class="eda-text-box">
        يتم اختيار <span class="eda-highlight">صورة عشوائية</span> من مجلد التدريب
        لعرض تأثير الفلاتر بشكل بصري، دون الحاجة إلى تطبيق المعالجة
        على كامل البيانات في هذه المرحلة.
    </div>
    <div class="eda-text-box">
        بعد قراءة الصورة وتحويلها من نظام
        <b>BGR</b> إلى <b>RGB</b>، يتم تطبيق نوعين مختلفين من فلاتر إزالة الضوضاء:
        <ul style="margin-right: 20px;">
            <li>
                <b>Median Filter</b>:
                فعال في إزالة الضوضاء النقطية (Salt & Pepper)
                مع الحفاظ على الحواف الأساسية.
            </li>
            <li>
                <b>Bilateral Filter</b>:
                يقلل الضوضاء مع الحفاظ على الحواف والتفاصيل الدقيقة،
                ويُعد مناسبًا للصور الطبية أو البيولوجية.
            </li>
        </ul>
    </div>
    <div class="eda-text-box">
        يتم بعد ذلك <span class="eda-highlight">عرض الصور جنبًا إلى جنب</span>
        (الصورة الأصلية، بعد Median Filter، بعد Bilateral Filter)
        باستخدام <b>Plotly</b> لتمكين المقارنة التفاعلية
        وتقييم أي الفلاتر أنسب للبيانات.
    </div>

</div>

<div class="eda-note">
    ℹ️ اختيار فلتر إزالة الضوضاء المناسب يؤثر مباشرة على وضوح الحواف
    ودقة الميزات التي سيتعلمها النموذج لاحقًا.
</div>

</div>


```python
# =========================
# Step 7: Image Denoising (FIXED)
# =========================

import cv2
import random
import numpy as np
from pathlib import Path
import plotly.express as px

TRAIN_PATH = Path("datasat orginal/Train")

# اختيار صورة عشوائية
class_dir = random.choice([d for d in TRAIN_PATH.iterdir() if d.is_dir()])
img_path = random.choice(list(class_dir.iterdir()))

# قراءة الصورة
img = cv2.imread(str(img_path))
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# تطبيق الفلاتر
median_denoised = cv2.medianBlur(img, 5)
bilateral_denoised = cv2.bilateralFilter(img, d=9, sigmaColor=75, sigmaSpace=75)

# دمج الصور للعرض
images = np.stack([img, median_denoised, bilateral_denoised])

# العرض التفاعلي
fig = px.imshow(
    images,
    facet_col=0,
    title="Step 7: Denoising (Original | Median | Bilateral)"
)

fig.update_layout(height=400)
fig.show()
```

<div class="eda-intro">

<div class="eda-container">
    <div class="eda-title">🌗 Step 8: تحسين التباين باستخدام CLAHE</div>
    <div class="eda-text-box">
        تهدف هذه الخلية إلى تحسين <span class="eda-highlight">التباين (Contrast)</span> في الصور
        باستخدام تقنية <b>CLAHE</b>  
        (<i>Contrast Limited Adaptive Histogram Equalization</i>).
    </div>
    <div class="eda-text-box">
        يتم اختيار <span class="eda-highlight">صورة عشوائية</span> من بيانات التدريب
        لعرض تأثير تحسين التباين بشكل بصري،
        دون تطبيق المعالجة على كامل مجموعة البيانات في هذه المرحلة.
    </div>
    <div class="eda-text-box">
        يتم تحويل الصورة من نظام <b>RGB</b> إلى نظام <b>LAB</b>،
        حيث تمثل قناة <b>L</b> شدة الإضاءة (Lightness)،
        مما يسمح بتحسين التباين دون التأثير على الألوان الأصلية.
    </div>
    <div class="eda-text-box">
        بعد فصل القنوات، يتم تطبيق <span class="eda-highlight">CLAHE</span> على قناة <b>L</b> فقط،
        مع استخدام:
        <ul style="margin-right: 20px;">
            <li><b>clipLimit</b> للتحكم في منع التضخيم الزائد للتباين</li>
            <li><b>tileGridSize</b> لتقسيم الصورة إلى مناطق صغيرة ومعالجتها محليًا</li>
        </ul>
        ثم يتم دمج القنوات مرة أخرى وتحويل الصورة إلى RGB.
    </div>
    <div class="eda-text-box">
        في النهاية، يتم عرض <span class="eda-highlight">الصورة قبل وبعد تحسين التباين</span>
        جنبًا إلى جنب باستخدام عرض تفاعلي،
        مما يسمح بتقييم مدى وضوح التفاصيل الدقيقة بعد المعالجة.
    </div>

</div>

<div class="eda-note">
    ℹ️ CLAHE مفيد بشكل خاص للصور ذات الإضاءة غير المتجانسة،
    ويُستخدم كثيرًا في الصور الطبية والبيولوجية.
</div>

</div>


```python
# =========================
# Step 8: Contrast Enhancement with CLAHE
# =========================

import cv2
import random
import numpy as np
from pathlib import Path
import plotly.express as px

TRAIN_PATH = Path("datasat orginal/Train")

# اختيار صورة عشوائية
class_dir = random.choice([d for d in TRAIN_PATH.iterdir() if d.is_dir()])
img_path = random.choice(list(class_dir.iterdir()))

# قراءة الصورة
img = cv2.imread(str(img_path))
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# تحويل الصورة إلى LAB لتطبيق CLAHE على قناة L
lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB)
l, a, b = cv2.split(lab)

# إنشاء CLAHE object
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
l_clahe = clahe.apply(l)

# دمج القنوات مرة أخرى وتحويل RGB
lab_clahe = cv2.merge([l_clahe, a, b])
img_clahe = cv2.cvtColor(lab_clahe, cv2.COLOR_LAB2RGB)

# عرض الصورة قبل / بعد
images = np.stack([img, img_clahe])
fig = px.imshow(images, facet_col=0, title="Step 8: CLAHE Contrast Enhancement")
fig.update_layout(height=400)
fig.show()
```

<div class="eda-intro">

<div class="eda-container">
    <div class="eda-title">🎚️ Step 9: تهيئة القيم اللونية (Color Normalization)</div>
    <div class="eda-text-box">
        تهدف هذه الخلية إلى تنفيذ خطوة <span class="eda-highlight">تهيئة القيم اللونية (Normalization)</span>
        من خلال تحويل قيم البكسلات إلى نطاق عددي موحّد
        مناسب لاستخدامه داخل نماذج التعلم العميق.
    </div>
    <div class="eda-text-box">
        في الصور الرقمية تكون قيم البكسلات عادةً بين
        <b>0 و 255</b>، لذلك يتم تحويلها إلى النطاق
        <span class="eda-highlight">0 → 1</span>
        عن طريق القسمة على 255 بعد تحويل نوع البيانات إلى
        <b>float32</b>.
    </div>
    <div class="eda-text-box">
        هذه الخطوة ضرورية لأنها:
        <ul style="margin-right: 20px;">
            <li>تُحسّن استقرار عملية التدريب</li>
            <li>تُسرّع تقارب النموذج (Faster Convergence)</li>
            <li>تمنع سيطرة القيم الكبيرة على التدرجات</li>
            <li>تجعل سلوك النموذج أكثر قابلية للتوقع</li>
        </ul>
    </div>
    <div class="eda-text-box">
        لأغراض العرض فقط، يتم <span class="eda-highlight">إعادة تحويل القيم</span>
        إلى النطاق 0–255 حتى يتمكن Plotly من عرض الصورة،
        مع التأكيد أن البيانات الفعلية المستخدمة في التدريب
        تكون بالقيم المُهيّأة (0–1).
    </div>

</div>

<div class="eda-note">
    ℹ️ Normalization خطوة أساسية لا يمكن تجاهلها،
    وتطبيقها بشكل صحيح يؤثر مباشرة على دقة واستقرار النموذج.
</div>

</div>


```python
# =========================
# Step 9: Color Normalization / Scaling
# =========================

import cv2
import random
import numpy as np
from pathlib import Path
import plotly.express as px

TRAIN_PATH = Path("datasat orginal/Train")

# اختيار صورة عشوائية
class_dir = random.choice([d for d in TRAIN_PATH.iterdir() if d.is_dir()])
img_path = random.choice(list(class_dir.iterdir()))

# قراءة الصورة
img = cv2.imread(str(img_path))
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# ---------------------
# Normalization
# ---------------------
# تحويل القيم من 0-255 إلى 0-1
img_normalized = img.astype(np.float32) / 255.0

# عرض الصورة قبل / بعد (Plotly يتطلب إعادة تحويل إلى 0-255 للعرض)
images_to_show = np.stack([img, (img_normalized*255).astype(np.uint8)])
fig = px.imshow(images_to_show, facet_col=0, title="Step 9: Normalization (0-1)")
fig.update_layout(height=400)
fig.show()
```

```python
# =========================
# Step 10: Edge Enhancement + Background Removal
# =========================

import cv2
import random
import numpy as np
from pathlib import Path
import plotly.express as px

TRAIN_PATH = Path("datasat orginal/Train")

# اختيار صورة عشوائية
class_dir = random.choice([d for d in TRAIN_PATH.iterdir() if d.is_dir()])
img_path = random.choice(list(class_dir.iterdir()))

# قراءة الصورة
img = cv2.imread(str(img_path))
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# ---------------------
# تحسين الحدود أولًا (Sharpening)
# ---------------------
kernel = np.array([[0,-1,0], [-1,5,-1], [0,-1,0]])
img_sharpened = cv2.filter2D(img, -1, kernel)

# ---------------------
# إزالة الخلفية بعد التوضيح
# ---------------------
gray = cv2.cvtColor(img_sharpened, cv2.COLOR_RGB2GRAY)
blur = cv2.GaussianBlur(gray, (5,5), 0)
_, mask = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
mask_rgb = cv2.merge([mask, mask, mask])
img_bg_removed = cv2.bitwise_and(img_sharpened, mask_rgb)

# ---------------------
# عرض النتائج
# ---------------------
images = np.stack([img, img_sharpened, img_bg_removed])
fig = px.imshow(images, facet_col=0, title="Original | Sharpened | Background Removed")
fig.update_layout(height=400)
fig.show()
```

```python
# =========================
# Step 11: Full Preprocessing Comparison
# =========================

import cv2
import random
import numpy as np
from pathlib import Path
import plotly.express as px

TRAIN_PATH = Path("datasat orginal/Train")
TARGET_SIZE = (300, 300)

# اختيار صورة عشوائية
class_dir = random.choice([d for d in TRAIN_PATH.iterdir() if d.is_dir()])
img_path = random.choice(list(class_dir.iterdir()))

# ---------------------
# 1. Original
# ---------------------
img_original = cv2.imread(str(img_path))
img_original = cv2.cvtColor(img_original, cv2.COLOR_BGR2RGB)

# ---------------------
# 2. Resize
# ---------------------
img_resized = cv2.resize(img_original, TARGET_SIZE)

# ---------------------
# 3. Denoising
# ---------------------
median_denoised = cv2.medianBlur(img_resized, 5)
bilateral_denoised = cv2.bilateralFilter(img_resized, d=9, sigmaColor=75, sigmaSpace=75)
# اختيار Bilateral كأفضل
img_denoised = bilateral_denoised

# ---------------------
# 4. Contrast Enhancement (CLAHE)
# ---------------------
lab = cv2.cvtColor(img_denoised, cv2.COLOR_RGB2LAB)
l, a, b = cv2.split(lab)
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
l_clahe = clahe.apply(l)
lab_clahe = cv2.merge([l_clahe, a, b])
img_clahe = cv2.cvtColor(lab_clahe, cv2.COLOR_LAB2RGB)

# ---------------------
# 5. Color Correction (Gray World)
# ---------------------
avgR = np.mean(img_clahe[:,:,0])
avgG = np.mean(img_clahe[:,:,1])
avgB = np.mean(img_clahe[:,:,2])
avgGray = (avgR + avgG + avgB) / 3
img_color_corrected = img_clahe.astype(np.float32)
img_color_corrected[:,:,0] = np.clip(img_color_corrected[:,:,0] * (avgGray/avgR), 0, 255)
img_color_corrected[:,:,1] = np.clip(img_color_corrected[:,:,1] * (avgGray/avgG), 0, 255)
img_color_corrected[:,:,2] = np.clip(img_color_corrected[:,:,2] * (avgGray/avgB), 0, 255)
img_color_corrected = img_color_corrected.astype(np.uint8)

# ---------------------
# 6. Sharpening
# ---------------------
kernel = np.array([[0,-1,0], [-1,5,-1], [0,-1,0]])
img_sharpened = cv2.filter2D(img_color_corrected, -1, kernel)

# ---------------------
# 7. Background Removal
# ---------------------
gray = cv2.cvtColor(img_sharpened, cv2.COLOR_RGB2GRAY)
blur = cv2.GaussianBlur(gray, (5,5), 0)
_, mask = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
mask_rgb = cv2.merge([mask, mask, mask])
img_bg_removed = cv2.bitwise_and(img_sharpened, mask_rgb)

# ---------------------
# عرض جميع النتائج
# ---------------------
images = np.stack([
    img_original, 
    img_resized, 
    img_denoised, 
    img_clahe, 
    img_color_corrected, 
    img_sharpened, 
    img_bg_removed
])

titles = [
    "Original", 
    "Resized", 
    "Denoised", 
    "CLAHE", 
    "Color Corrected", 
    "Sharpened", 
    "Background Removed"
]

fig = px.imshow(images, facet_col=0, title="Full Preprocessing Pipeline")
fig.update_layout(height=400)
fig.show()
```

```python
# =========================
# Step 12: Data Augmentation (TF only)
# =========================

import cv2
import random
import numpy as np
from pathlib import Path
import tensorflow as tf
import plotly.express as px

TRAIN_PATH = Path("datasat orginal/Train")
TARGET_SIZE = (300, 300)

# اختيار صورة عشوائية
class_dir = random.choice([d for d in TRAIN_PATH.iterdir() if d.is_dir()])
img_path = random.choice(list(class_dir.iterdir()))

# قراءة الصورة وتحويلها float32
img = cv2.imread(str(img_path))
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img = cv2.resize(img, TARGET_SIZE)
img = img.astype(np.float32) / 255.0

# تحويل الصورة إلى Tensor
img_tf = tf.convert_to_tensor(img)

# ---------------------
# دالة Augmentation
# ---------------------
def augment_image(image):
    # Flip أفقي وعمودي
    image = tf.image.random_flip_left_right(image)
    image = tf.image.random_flip_up_down(image)
    
    # Rotation 90°, 180°, 270°
    k = tf.random.uniform([], minval=0, maxval=4, dtype=tf.int32)
    image = tf.image.rot90(image, k=k)
    
    # Brightness & Contrast
    image = tf.image.random_brightness(image, max_delta=0.2)
    image = tf.image.random_contrast(image, lower=0.8, upper=1.2)
    
    # Hue & Saturation
    image = tf.image.random_hue(image, max_delta=0.05)
    image = tf.image.random_saturation(image, lower=0.8, upper=1.2)
    
    return tf.clip_by_value(image, 0.0, 1.0)

# ---------------------
# إنشاء عدة عينات Augmented
# ---------------------
augmented_images = [img]  # الأصلية
for _ in range(5):
    augmented = augment_image(img_tf).numpy()
    augmented_images.append(augmented)

# ---------------------
# عرض النتائج
# ---------------------
images = np.stack(augmented_images)
fig = px.imshow(images, facet_col=0, title="Step 12: TF Augmentation Examples")
fig.update_layout(height=400)
fig.show()
```

```python
# =========================
# Step 12b: Pixel Statistics (Interactive Histogram)
# =========================

import cv2
import numpy as np
from pathlib import Path
import plotly.graph_objects as go
import random

TRAIN_PATH = Path("datasat orginal/Train")
TARGET_SIZE = (300, 300)

# اختيار صورة عشوائية
class_dir = random.choice([d for d in TRAIN_PATH.iterdir() if d.is_dir()])
img_path = random.choice(list(class_dir.iterdir()))

# قراءة الصورة وتحويلها إلى RGB + Resize
img_original = cv2.imread(str(img_path))
img_original = cv2.cvtColor(img_original, cv2.COLOR_BGR2RGB)
img_resized = cv2.resize(img_original, TARGET_SIZE)

# مثال: بعد Denoising + CLAHE + Color Correction (يمكن تعديل حسب رغبتك)
import cv2
# Denoising Bilateral
img_denoised = cv2.bilateralFilter(img_resized, d=9, sigmaColor=75, sigmaSpace=75)

# CLAHE
lab = cv2.cvtColor(img_denoised, cv2.COLOR_RGB2LAB)
l, a, b = cv2.split(lab)
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
l_clahe = clahe.apply(l)
img_clahe = cv2.cvtColor(cv2.merge([l_clahe, a, b]), cv2.COLOR_LAB2RGB)

# Color Correction (Gray World)
avgR, avgG, avgB = np.mean(img_clahe[:,:,0]), np.mean(img_clahe[:,:,1]), np.mean(img_clahe[:,:,2])
avgGray = (avgR + avgG + avgB)/3
img_color_corrected = img_clahe.astype(np.float32)
img_color_corrected[:,:,0] = np.clip(img_color_corrected[:,:,0] * (avgGray/avgR),0,255)
img_color_corrected[:,:,1] = np.clip(img_color_corrected[:,:,1] * (avgGray/avgG),0,255)
img_color_corrected[:,:,2] = np.clip(img_color_corrected[:,:,2] * (avgGray/avgB),0,255)
img_color_corrected = img_color_corrected.astype(np.uint8)

# ---------------------
# رسم هيستوجرام تفاعلي لكل قناة قبل وبعد
# ---------------------
fig = go.Figure()

# Original
for i, color in enumerate(['Red','Green','Blue']):
    fig.add_trace(go.Histogram(
        x=img_resized[:,:,i].flatten(),
        name=f'Original {color}',
        opacity=0.6
    ))

# After preprocessing
for i, color in enumerate(['Red','Green','Blue']):
    fig.add_trace(go.Histogram(
        x=img_color_corrected[:,:,i].flatten(),
        name=f'Processed {color}',
        opacity=0.6
    ))

fig.update_layout(
    barmode='overlay',
    title="Pixel Value Distribution: Original vs Processed",
    xaxis_title="Pixel Value (0-255)",
    yaxis_title="Frequency",
    height=500
)
fig.show()
```

```python
# ====== A: Readiness check (sanity + model forward pass) ======
# شغّل هذه الخلية للتحقق أن الصور تُحمَل وتُعالَج وتدخل النموذج بدون أخطاء.

import os
from pathlib import Path
import numpy as np
import json
import datetime
import sys

DATA_ROOT = Path("datasat orginal")
TRAIN_DIR = DATA_ROOT / "Train"
IMAGE_SIZE = (300, 300)
BATCH_SIZE = 8

print("Dataset root:", DATA_ROOT)
assert TRAIN_DIR.exists(), f"لم أجد مسار التدريب: {TRAIN_DIR}"

# جمع مسارات الصور والفئات
image_exts = {".jpg", ".jpeg", ".png", ".bmp"}
paths = []
labels = []
classes = sorted([d.name for d in TRAIN_DIR.iterdir() if d.is_dir()])
if not classes:
    raise SystemExit("لا توجد مجلدات فئات داخل Train — تأكد من تنظيم المجلدات بالفئات.")

for ci, cls in enumerate(classes):
    for p in (TRAIN_DIR/cls).iterdir():
        if p.suffix.lower() in image_exts:
            paths.append(str(p))
            labels.append(ci)

print(f"Found {len(paths)} images across {len(classes)} classes.")
print("Classes:", classes)

# حاول استخدام TensorFlow لإعداد tf.data pipeline واختبار دفعة + forward pass
try:
    import tensorflow as tf
    from tensorflow.keras.applications import efficientnet
    print("TensorFlow version:", tf.__version__)
except Exception as e:
    print("TensorFlow غير متوفر أو حدث خطأ في الاستيراد:", e)
    tf = None

def preprocess_path_label(path, label):
    # يقرأ، يغير الحجم إلى IMAGE_SIZE، ويُعيد normalised float32 [0,1]
    img = tf.io.read_file(path)
    img = tf.image.decode_image(img, channels=3, expand_animations=False)
    img = tf.image.convert_image_dtype(img, tf.float32)            # to [0,1]
    img = tf.image.resize(img, IMAGE_SIZE)
    return img, label

if tf is not None:
    try:
        ds = tf.data.Dataset.from_tensor_slices((paths, labels))
        ds = ds.shuffle(2048).map(preprocess_path_label, num_parallel_calls=tf.data.AUTOTUNE)
        ds = ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
        it = iter(ds)
        imgs_batch, labels_batch = next(it)
        print("Batch shapes:", imgs_batch.shape, labels_batch.shape)
        print("dtypes:", imgs_batch.dtype, labels_batch.dtype)
        print("Min/Max in batch:", float(tf.reduce_min(imgs_batch)), float(tf.reduce_max(imgs_batch)))
        # فحص وجود NaNs
        has_nan = tf.math.reduce_any(tf.math.is_nan(imgs_batch))
        print("Any NaNs in batch?", bool(has_nan.numpy()))
    except Exception as e:
        print("خطأ أثناء إنشاء/جلب دفعة من tf.data:", e)
        imgs_batch = None

    # Forward pass عبر EfficientNetB3 (weights=None لاستخدام نموذج بسيط للاختبار)
    if imgs_batch is not None:
        try:
            model = efficientnet.EfficientNetB3(weights=None, include_top=False, input_shape=(IMAGE_SIZE[0], IMAGE_SIZE[1], 3))
            out = model(imgs_batch, training=False)
            print("Model forward pass OK. Output shape:", out.shape)
        except Exception as e:
            print("خطأ أثناء forward pass عبر EfficientNetB3:", e)
else:
    print("تخطّي اختبار النموذج لأن TensorFlow غير مثبت.")

print("✅ Readiness check finished.")
```
